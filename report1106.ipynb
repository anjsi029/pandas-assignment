## 문제 7. loc [조건, 열]

- 12살 이하 그리고 3등석 승객의 age , sex , survived 열만 kids 변수에 담고, 행 수와 생존률 (survived 평균, 소수 둘째 자리)을 출력하세요.

```python
mask = (titanic["age"] <= 12) & (titanic["class"] == "Third")
kids = titanic[mask][["age","sex","survived"]]
print("length:", len(kids))
result = kids["survived"].mean()
print("survived:", round(result , 2))
```

## 10. 종합 — 20~40세 남성, 1·2등석 vs 3등석
질문 : "20~40세 남성 승객의 생존률은 객실 등급에 따라 얼마나 다를까?"

1. 남성 · 나이 20~40세(between) · 1·2등석(isin) 조건으로 pclass, age, fare, survived 열을 선택하세요.

```python
mask = (titanic["sex"]=="male") & (titanic["age"].between(20,40)) & (titanic["class"].isin(["First" , "Second"]))
result = titanic[mask][["pclass","age","fare","survived"]]
result
```

2. 행 수, 생존률(소수 둘째 자리), 요금 상위 3명을 출력하세요.

```python
print("length:", len(result))
print("survived:", round(result["survived"].mean(),2))
result.sort_values(by="fare",ascending=False).head(3)
```

3. 같은 조건을 query로 써서 행 수를 검증하세요.

```python
titanic.query('sex=="male" and 20 <= age <= 40 and `class` in ["First" , "Second"]')
```

4. 등급만 3등석으로 바꿔 생존률을 구하고, 두 생존률을 비교하는 해석을 마크다운 셀에 2~3문장으로 쓰세요.

```python
result = titanic.query('sex=="male" and 20 <= age <= 40 and `class` == "Third"')
target = result["survived"].mean()
round(target,2)
```

- 1 ~ 2등석 승객의 생존률은 24% 이고, 3등석 승객의 생존률은 15% 로 1,2등석 승객보다 낮다. 값비싼 티켓을 구매할 수 있는 사회적 능력과 생존률이 연관되어 있음을 확인할 수 있다.
